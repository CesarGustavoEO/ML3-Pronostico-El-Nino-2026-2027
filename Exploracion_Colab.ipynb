```python
from google.colab import files
from pathlib import Path
import zipfile, io
print('Selecciona el ZIP individual de este módulo:')
uploaded = files.upload()
zip_names = [n for n in uploaded if n.lower().endswith('.zip')]
if len(zip_names) != 1:
    raise ValueError('Sube exactamente un ZIP de este módulo.')
with zipfile.ZipFile(io.BytesIO(uploaded[zip_names[0]])) as z:
    z.extractall('/content/entrega')
base = Path('/content/entrega') / 'ML3_Pronostico_Pacifico'
if not (base / 'app.py').exists():
    raise FileNotFoundError('No se encontró app.py en el ZIP esperado.')
print('Archivos encontrados:', base)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
plt.rcParams.update({"figure.figsize":(10,5),"font.size":11})
```

```python
hist = pd.read_csv(base/'datos'/'enso_dataset_maestro.csv',parse_dates=['fecha'])
pron = pd.read_csv(base/'datos'/'03_PRONOSTICO_CONGELADO.csv')
met = pd.read_csv(base/'datos'/'02_metricas_por_horizonte.csv')
display(pron)
display(met)
q=hist.sort_values('fecha').tail(18)
x=pd.to_datetime(pron.fecha)
fig,ax=plt.subplots(figsize=(11,5))
ax.plot(q.fecha,q.anom_nino34,color='#087E8B',linewidth=3,label='Observado NOAA')
ax.plot(pd.concat([q.fecha.tail(1),pd.Series(x)]),[q.anom_nino34.iloc[-1]]+pron.pronostico_nino34_c.tolist(),color='#E58A36',linestyle='--',marker='o',label='Estimación ML')
ax.fill_between(x,pron.limite_inferior_c,pron.limite_superior_c,color='#E58A36',alpha=.17,label='Margen histórico orientativo')
ax.axhline(0,color='gray',linewidth=1)
ax.set_ylabel('Anomalía Niño 3.4 (°C)'); ax.set_title('ML 3 | Observado y estimado')
ax.legend(); ax.grid(alpha=.2); plt.xticks(rotation=25); plt.tight_layout(); plt.show()
```
